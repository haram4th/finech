```python
data = response_body.decode('utf-8')
data
```

```python
import json
```

```python
data = json.loads(data)
data
```

```python
len(data['items'])
```

```python
data.keys()
```

```python
def text_clean(x):
    result = x.replace("<b>", "").replace("</b>", "")
    return result
```

```python
result = {}
for item in data['items']:
    for key, value in item.items():
        value = text_clean(value)
        result.setdefault(key, []).append(value)
result
```
