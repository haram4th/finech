```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

options = Options()
options.add_experimental_option("detach", True)
options.add_argument("start-maximized")
options.add_argument("Chrome/135.0.0.0")
options.add_argument("lang=ko_KR")


driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
    )
driver.get("https://cafe.naver.com/fujipeople")
```

```python
#로그인 버튼 클릭
driver.find_element(By.ID, "gnb_login_button").click()
```

```python
# 질문 게시판 클릭
driver.find_element(By.ID, "menuLink51").click()
```

```python
# 질문 글 목록 게시판글
question_board_tbodys = driver.find_elements(By.CSS_SELECTOR, "div.article-board table.article-table tbody")
third_tbody = question_board_tbodys[2]
```

```python
qeustion_lists = third_tbody.find_elements(By.TAG_NAME, "tr")
```

```python
qeustion_lists[0].find_element(By.CSS_SELECTOR, "a.article").text
```

```python
# 질문 
qeustion_lists[0].find_element(By.CSS_SELECTOR, "a.article").click()
# 카페 ifram으로 진입
driver.switch_to.frame("cafe_main")
```

```python
# 질문 제목
question_title = driver.find_element(By.CSS_SELECTOR, ".title_text").text
# 질문 본문 내용
question_content = driver.find_element(By.CSS_SELECTOR, ".content.CafeViewer").text
# 댓글 리스트
replies = driver.find_elements(By.CSS_SELECTOR, "div.CommentBox > div:nth-child(2) > ul > li")
reply_list = []
for reply in replies:
    reply_text = reply.find_element(By.CSS_SELECTOR, "div.comment_text_box > p > span").text
    reply_list.append(reply_text)
```

```python
print(question_title, end="\n\n")
print(question_content, end="\n\n")
print(reply_list)
```
