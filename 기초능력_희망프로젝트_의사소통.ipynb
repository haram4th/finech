```python
result1 = df1.join(df2['1지망 프로젝트를 선택해 주세요'])
result1
```

```python
result = result1.join(df3)
result
```

```python
result = result.reset_index()
result
```

```python
result = result.rename(columns={'Unnamed: 0': '이름'})
result
```

```python
del result['타임스탬프']
```

```python
result
```

```python
main_cols = []
extra_cols = []
for col in result.columns:
    if "기타" in col:
        extra_cols.append(col)
    else:
        main_cols.append(col)
```

```python
main_cols
```

```python
extra_cols
```

```python
front = main_cols[5:10]
```

```python
backend = main_cols[10:15]
```

```python
database = main_cols[15:18]
```

```python
python_basic = main_cols[18:21]
```

```python
web_scraping = main_cols[21:26]
```

```python
ml = main_cols[26:32]
```

```python
dl = main_cols[32:37]
```

```python
cloud = main_cols[38:]
```

```python
category = [front, backend, database, python_basic, web_scraping, ml, dl,  cloud]
category_name = ['front', 'backend', 'database', 'python_basic', 'web_scraping', 'ml', 'dl',  'cloud']
```

```python
main_df = result[main_cols]
main_df
```

```python
category
```

```python
for cate, cate_name in zip(category, category_name):
    max_point = len(cate) * 5
    point = main_df[cate].sum(axis=1)
    main_df[cate_name] = ((point / max_point) * 100).round().astype(int)
```

```python
main_df
```

```python
extra_df = result[extra_cols]
extra_df
```

```python
final_df = pd.concat([main_df, extra_df], axis=1)
final_df
```

```python
final_df.columns
```

```python
data = final_df[['이름', '의사소통', '기획력', '커뮤니케이션', '1지망 프로젝트를 선택해 주세요'] + ['front', 'backend',
       'database', 'python_basic', 'web_scraping', 'ml', 'dl', 'cloud',
       '기타 프론트엔드 관련 기술을 보유하고 있다면 적어주세요.',
       '기타 보유한 백엔드 기술을 적어주세요. java spring 등',
       '기타 데이터베이스 관련 기술을 보유하고 있다면 적어주세요.', '기타 사용 가능한 라이브러리등을 적어주세요',
       '기타 보유한 데이터 수집 관련 기술을 적어주세요.', '기타 보유한 머신러닝 관련 기술을 적어주세요.',
       '기타 보유한 클라우드 기술을 적어주세요.']]
data
```

```python
data.columns
```

```python
skill_data = data[['이름', '1지망 프로젝트를 선택해 주세요', '의사소통', '기획력', '커뮤니케이션', 'front', 'backend',
       'database', 'python_basic', 'web_scraping', 'ml', 'dl', 'cloud']]
skill_data
```

```python
skill_data[['의사소통', '기획력', '커뮤니케이션']] = skill_data[['의사소통', '기획력', '커뮤니케이션']] / 5 * 100
```

```python
skill_data
```

```python
numeric_cols = skill_data.select_dtypes(include="number").columns
```

```python
skill_data_sorted = skill_data.sort_values(by="1지망 프로젝트를 선택해 주세요")
skill_data_sorted.to_csv("./프로젝트_참여자_스킬비교_personal.csv", encoding='utf-8-sig')
skill_data_sorted
```

```python
skill_data_group = skill_data.groupby('1지망 프로젝트를 선택해 주세요')[numeric_cols].mean()
```

```python
skill_data_group = skill_data_group.astype(int)
skill_data_group
```

```python
skill_data_group.to_csv("./프로젝트_참여자_스킬비교.csv", encoding='utf-8-sig')
```

```python
# 인덱스를 문자열로 명확하게 변환
skill_data_group.index = skill_data_group.index.astype(str)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 예시로 skill_data_group가 이미 준비되어 있다고 가정합니다.
# 만약 준비되지 않았다면 아래 코드를 사용해 생성하세요:
# skill_data_group = skill_data.groupby("1지망 프로젝트를 선택해 주세요").mean(numeric_only=True)
# skill_data_group.index = skill_data_group.index.astype(str)

# DataFrame melt
df_melted = skill_data_group.reset_index().melt(id_vars="1지망 프로젝트를 선택해 주세요",
                                                var_name="스킬 항목",
                                                value_name="평균 점수")

plt.figure(figsize=(12, 6))
sns.barplot(data=df_melted,
            x="스킬 항목",
            y="평균 점수",
            hue="1지망 프로젝트를 선택해 주세요")

plt.title("프로젝트별 스킬 평균 비교")
plt.xlabel("스킬 항목")
plt.ylabel("평균 점수 (퍼센트/5점 기준)")
plt.xticks(rotation=45, ha='right')
plt.legend(title="프로젝트", loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```
