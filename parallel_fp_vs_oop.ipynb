## 1. 함수형 프로그래밍 + 병렬 처리

```python
import pandas as pd
from multiprocessing import Pool
from itertools import chain

# 순수 함수로 페이지 처리
def process_page(page):
    return [
        dict(zip(page['header']['columns'], item.values()))
        for item in page['body']['items']
    ]

def flatten_data_functional_parallel(data, num_workers=4):
    with Pool(num_workers) as pool:
        results = pool.map(process_page, data)

    # 결과 병합
    flattened = list(chain.from_iterable(results))
    return pd.DataFrame(flattened)

# 사용 예시 (if __name__ == "__main__": 블록 안에서 실행해야 함)
df_functional_parallel = flatten_data_functional_parallel(data)
```

## 2. 객체지향 프로그래밍 + 병렬 처리

```python
import pandas as pd
from multiprocessing import Pool

# 외부 함수로 페이지 처리
def process_page_oop(page):
    columns = page['header']['columns']
    return [
        dict(zip(columns, item.values()))
        for item in page['body']['items']
    ]

class DataFlattenerParallel:
    def __init__(self, data):
        self.data = data
        self.records = []

    def process_parallel(self, num_workers=4):
        with Pool(num_workers) as pool:
            results = pool.map(process_page_oop, self.data)

        for page_records in results:
            self.records.extend(page_records)

        return self

    def to_dataframe(self):
        return pd.DataFrame(self.records)

# 사용 예시
# flattener = DataFlattenerParallel(data).process_parallel()
# df_oop_parallel = flattener.to_dataframe()
```
