# 리스트 표현식
* 리스트 안에 for 반복문과 if else 조건문을 한 번에 적어서 리스트를 만드는 방법
* 여러 줄의 코드를 1줄로 줄일 수 있다.
* for문과 if else문을 사용해서 처리할 때 보다 속도가 빠르다.
* [실행할 명령 for 변수 in list, tuple]

```python
for i in range(1, 11):
    print(f"{i}는 짝수입니다.") if i % 2 == 0 else print(f"{i}는 홀수입니다.")
```

* 1부터 10까지 들어있는 리스트 만들기

```python
l1 = []
for i in range(1, 11):
    l1.append(i)
l1
```

```python
[i for i in range(1, 11)]
```

## with open
* with open은 close()를 하지 않아도 자동으로 close()처리를 해줌

```python
with open("test2.txt", 'a') as file3:
    file3.write("\nwith open을 사용하면 파일을 자동으로 닫아주어 편리하다.")
```

```python
with open("test2.txt", 'r') as file4:
#     data = file4.readlines()
    data = file4.read()
print(data)
```

# encoding에 따른 읽기 오류
* 운영체제나(windows, linux, macOS), 한글/영어 윈도우에 따라서 인코딩 되는 포멧이 달라짐
* 인코딩에 따른 오류가 발생 가능
* web표준은 utf-8이므로 인코딩은 utf-8로 하는 것이 좋다.
* windows의 경우 인코딩이 cp949이므로 주의

```python
with open("test_utf-8.txt", 'w', encoding='utf-8') as file5:
    file5.write("이것은 with open으로 만들 utf-8 인코딩 파일입니다.")
```

```python
with open("test_utf-8.txt", 'r') as file6:
    file6_data = file6.read()
print(file6_data)
```

```python
with open("test_utf-8.txt", 'r', encoding="utf-8") as file6:
    file6_data = file6.read()
print(file6_data)
```
