* 매개변수 3개짜리 함수 
* num_maker라는 함수를 매개변수 3개를 이용해서 만드세요.
* 숫자를 3개 입력받아 첫번째 숫자는 시작숫자, 두번째 숫자는 끝숫자, 마지막 숫자는 step 으로 작동하게 하세요.
* 먼저 숫자를 생성한 결과를 출력하고
* 숫자 리스트에 * 10한 결과를 다시 출력하세요.
* num_maker(1, 11, 2)
* 결과: 1,3,5,7,9

```python
def num_maker(start, end, step):
    result = list(range(start, end, step))
    result2 = []
    print(result)
    for i in result:
        result2.append(i*10)
    print(result2)
    return result, result2
```

```python
num_maker(end=11, start=1, step=2)
```

* num_maker2 - 정방향, 역방향 리스트가 출력되는 함수

```python
def num_maker2(start, end, step):
    list_result = list(range(start, end, step))
    reversed_list = list_result[::-1]
#     print(list_result, reversed_list)
    return list_result, reversed_list
```

* 반복문에서 break 처럼 활용 가능
* 함수 내에서 return이 실행되면 함수는 종료

* 1에서 100까지 숫자중에서 짝수만 더해서 출력하시오.
* 단, 숫자가 50이상이 되면 멈추시오.
* 더해지는 과정도 출력

```python
num = list(range(1,101))

   
```

```python
def even(num):
#     print(num)
    hap = 0
    for i in num:
        if i < 50:
            if i % 2 == 0:
                hap += i
                print(hap, end=" ")
        else:
            break
    return hap
    
```

```python
even(list(range(1, 101)))
```

```python
def even2(num):
#     print(num)
    hap = 0
    for i in num:
        if i < 50:
            if i % 2 == 0:
                print(f"{i}+{hap}=", end="")
                hap += i
                print(f"{hap}", end=" ")
        else:
            return hap
    
```

* 위치인수와 가변인수 동시에 받기

* def muls(a, *args): OK
* def muls(*args, a): X

```python
# 정수와 리스트를 입력받아 리스의 요소에 입력받은 정수를 곱하는 함수
```

```python
def muls(a, *args):
    print(a, args)
    result = []
    for num in args:
        result.append(a*num)
    return result
```

```python
l2 = [1,2,3,4,5]
```

```python
# muls(6,1,2,3,4,5)
muls(6, *l2)
```

```python
def muls(*args, a):
    print(a, args)
    result = []
    for num in args:
        result.append(a*num)
    return result
```

```python
muls(1, 2, 3)
```

* 1에서 10까지 수 중에서 3의 배수만 문자열로 바꾸기

```python
def tostr(x):
    result=[]
    for i in x:
        if i % 3 ==0:
            result.append(str(i))
        else:
            result.append(i)
    return result
```

# 기본문법

```python
try:
    # 예외가 발생할 위험이 있는 코드
except Exception as e:
    print(e) # 어떤 에러인지 출력
    # 예외가 발생했을 때 실행할 코드
else:
    # 예외가 발생하지 않았을 때 실행할 코드
finally:
    # 예외가 발생하든지 안하든지 무조건 실행할 코드
```

* 숫자를 0나눌때 에러

```python
for i in range(5):
    num = int(input("숫자를 입력해 주세요"))
    try:
        x = 10 / num
    except Exception as e:
        print(e)
        print("숫자 0으로는 나눗셈을 할 수 없습니다.")
    else:
        print(x)
    finally:
        print("계산이 완료되었습니다.", end="\n\n")
        
```
